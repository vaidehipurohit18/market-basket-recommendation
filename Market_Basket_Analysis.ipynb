```python
# Import Required Libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Transaction Encoder converts transaction lists into matrix format
from mlxtend.preprocessing import TransactionEncoder

# Apriori algorithm for frequent itemsets
from mlxtend.frequent_patterns import apriori

# Association rules generation
from mlxtend.frequent_patterns import association_rules
```

```python
# Generate Large Realistic Transaction Dataset

transactions = []

# Grocery Patterns
for i in range(50):
    transactions.append([
        'Milk',
        'Bread',
        'Butter'
    ])

for i in range(40):
    transactions.append([
        'Milk',
        'Eggs',
        'Bread'
    ])

for i in range(35):
    transactions.append([
        'Bread',
        'Butter',
        'Jam'
    ])

for i in range(30):
    transactions.append([
        'Rice',
        'Dal',
        'Oil'
    ])

for i in range(25):
    transactions.append([
        'Tea',
        'Biscuits',
        'Sugar'
    ])

for i in range(25):
    transactions.append([
        'Coffee',
        'Milk',
        'Sugar'
    ])

for i in range(20):
    transactions.append([
        'Pizza',
        'Soda'
    ])

for i in range(20):
    transactions.append([
        'Burger',
        'Fries',
        'Soda'
    ])

for i in range(20):
    transactions.append([
        'Pasta',
        'Sauce',
        'Cheese'
    ])

for i in range(15):
    transactions.append([
        'Laptop',
        'Mouse',
        'Keyboard'
    ])

for i in range(15):
    transactions.append([
        'Phone',
        'Charger',
        'Earphones'
    ])

for i in range(15):
    transactions.append([
        'Notebook',
        'Pen',
        'Marker'
    ])

for i in range(12):
    transactions.append([
        'Soap',
        'Shampoo'
    ])

for i in range(12):
    transactions.append([
        'Detergent',
        'Fabric Softener'
    ])

for i in range(10):
    transactions.append([
        'Chocolate',
        'Ice Cream'
    ])

# Display Sample
transactions[:10]
```

```python
te = TransactionEncoder()

te_array = te.fit(transactions).transform(transactions)

basket_df = pd.DataFrame(
    te_array,
    columns=te.columns_
)
```

```python
# Generate Frequent Itemsets

frequent_items = apriori(
    basket_df,
    min_support=0.03,
    use_colnames=True
)

frequent_items = frequent_items.sort_values(
    by='support',
    ascending=False
)
```

```python
# Generate Association Rules

rules = association_rules(
    frequent_items,
    metric="confidence",
    min_threshold=0.5
)

# Display Rules
rules
```

```python
# Select Important Columns

rules = rules[[
    'antecedents',
    'consequents',
    'support',
    'confidence',
    'lift'
]]

# Sort by Lift
rules = rules.sort_values(
    by='lift',
    ascending=False
)

# Display Rules
rules
```

```python
# Top Frequent Items

top_items = frequent_items.sort_values(
    by='support',
    ascending=False
).head(10)

# Plot

plt.figure(figsize=(10,5))

sns.barplot(
    x=top_items['support'],
    y=top_items['itemsets'].astype(str)
)

plt.title("Top Frequent Itemsets")

plt.xlabel("Support")

plt.ylabel("Itemsets")

plt.show()
```

# Business Insights
